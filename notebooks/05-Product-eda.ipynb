### Imports

```python
import json
import pandas as pd
from pathlib import Path

DATA_DIR = Path("../data/raw")
```

### Load the data and investigate it

```python
all_rows = []

for path in DATA_DIR.glob("*/*.json"):
    with open(path, encoding="utf-8") as f:
        rows = json.load(f)
        
    all_rows.extend(rows)

df = pd.DataFrame(all_rows)

print("Total rows:", len(df))
print("Total columns:", len(df.columns))

df.head()
```

```python

df.info()

df.groupby("store").size()
```

```python

print("Total records:", len(df))

print(
    "Unique products:",
    df[["store", "id"]].drop_duplicates().shape[0]
)

print(
    "Duplicate records:",
    df.duplicated(subset=["store", "id"]).sum()
)
```

```python
duplicates = df[
    df.duplicated(subset=["store", "id"], keep=False)
].sort_values(["store", "id"])

display(
    duplicates[
        [
            "store",
            "id",
            "name",
            "price",
            "regular_price",
            "unit_price",
            "category"
        ]
    ]
)
```

```python

print("Exact duplicate records:", df.duplicated().sum())

print(
    "Duplicate product identifiers:",
    df.duplicated(subset=["store", "id"]).sum()
)

print("\nDuplicates by store:")

display(
    df[df.duplicated(subset=["store", "id"])]
    .groupby("store")
    .size()
)
```

```python

missing_by_store = df.groupby("store").agg(
    total_products=("id", "size"),
    missing_brand=("brand", lambda x: x.isna().sum()),
    missing_unit_price=("unit_price", lambda x: x.isna().sum()),
    missing_unit=("unit", lambda x: x.isna().sum()),
    missing_category=("category", lambda x: x.isna().sum())
)

display(missing_by_store)
```

```python

missing_unit_prices = df[df["unit_price"].isna()]

display(
    missing_unit_prices[
        ["store", "name", "price", "unit", "category"]
    ].head(30)
)
```

```python

display(
    missing_unit_prices
    .groupby(["store", "category"], dropna=False)
    .size()
    .sort_values(ascending=False)
    .head(20)
)
```

```python

display(
    df.groupby(["unit", "store"], dropna=False)
    .size()
    .unstack(fill_value=0)
)
```

### Work with data

produces a new clean dataframe df_clean

#### Remove duplicates

```python
df_clean = df.drop_duplicates().copy()
df_clean = df_clean.reset_index(drop=True)

print("Original records:", len(df))
print("Clean records:", len(df_clean))
print("Removed:", len(df) - len(df_clean))
```

#### Standardise measurement units

```python
df_clean["unit"] = df_clean["unit"].replace({
    "units": "pcs",
    "vnt": "pcs",
})

display(
    df_clean.groupby(["unit", "store"], dropna=False)
    .size()
    .unstack(fill_value=0)
)
```

### Continue missing prices investigation

```python
df_clean["main_category"] = (
    df_clean["category"]
    .str.split("/")
    .str[0]
    .str.strip("-")
)

display(
    df_clean.groupby(["store", "main_category"])
    .size()
)
```

```python

rimi_df = df_clean[df_clean["store"] == "rimi"]

missing_summary = rimi_df.groupby("main_category").agg(
    total=("id", "size"),
    missing_unit_price=("unit_price", lambda x: x.isna().sum())
)

missing_summary["missing_pct"] = (
    missing_summary["missing_unit_price"]
    / missing_summary["total"]
    * 100
).round(2)

display(
    missing_summary.sort_values(
        "missing_pct", ascending=False
    )
)
```

```python

display(
    df_clean[df_clean["category"].isna()][
        ["store", "id", "name", "price", "unit_price", "url"]
    ]
)
```

```python

print("Non-positive prices:",
      (df_clean["price"] <= 0).sum())

print("Non-positive regular prices:",
      (df_clean["regular_price"] <= 0).sum())

print("Non-positive unit prices:",
      (df_clean["unit_price"] <= 0).sum())

print("Promotions with inconsistent prices:",
      (
          df_clean["on_promo"]
          & (df_clean["price"] >= df_clean["regular_price"])
      ).sum())

print("Loyalty card required without promotion:",
      (
          df_clean["loyalty_card_required"]
          & ~df_clean["on_promo"]
      ).sum())
```

```python
mask = (
    (df_clean["store"] == "barbora")
    & df_clean["category"].isna()
    & df_clean["name"].str.contains("YOPRO", case=False, na=False)
)

print("Products to correct:", mask.sum())

df_clean.loc[mask, "main_category"] = (
    "pieno-gaminiai-kiausiniai-ir-majonezas"
)

print(
    "Remaining missing main categories:",
    df_clean["main_category"].isna().sum()
)
```

```python
invalid_prices = df_clean[
    (df_clean["price"] <= 0)
    | (df_clean["regular_price"] <= 0)
    | (df_clean["unit_price"] <= 0)
]

display(
    invalid_prices[
        [
            "store",
            "id",
            "name",
            "price",
            "regular_price",
            "unit_price",
            "unit",
            "url"
        ]
    ]
)
```

```python
invalid_promos = df_clean[
    df_clean["on_promo"]
    & (df_clean["price"] >= df_clean["regular_price"])
]

display(invalid_promos.groupby("store").size())

display(
    invalid_promos[
        [
            "store",
            "name",
            "price",
            "regular_price",
            "unit_price",
            "loyalty_card_required"
        ]
    ].head(30)
)
```

```python
loyalty_without_promo = df_clean[
    df_clean["loyalty_card_required"]
    & ~df_clean["on_promo"]
]

display(loyalty_without_promo.groupby("store").size())

display(
    loyalty_without_promo[
        [
            "store",
            "name",
            "price",
            "regular_price",
            "unit_price",
            "on_promo",
            "loyalty_card_required"
        ]
    ].head(30)
)
```

```python
promo_summary = df_clean.groupby("store").agg(
    total_products=("id", "size"),
    on_promo=("on_promo", "sum"),
    loyalty_required=("loyalty_card_required", "sum")
)

display(promo_summary)
```

```python
display(
    df_clean[
        (df_clean["store"] == "rimi")
        & df_clean["unit_price"].isna()
    ][
        ["id", "name", "price", "main_category", "url"]
    ].sample(10, random_state=42)
)
```

```python
invalid_price_mask = (
    (df_clean["price"] <= 0)
    | (df_clean["regular_price"] <= 0)
    | (df_clean["unit_price"] <= 0)
)

df_invalid = df_clean[invalid_price_mask].copy()

df_clean = df_clean[~invalid_price_mask].copy()
df_clean = df_clean.reset_index(drop=True)

print("Excluded products:", len(df_invalid))
print("Remaining products:", len(df_clean))
```

```python
rimi_mask = df_clean["store"] == "rimi"

df_clean["category_clean"] = df_clean["category"]

df_clean.loc[rimi_mask, "category_clean"] = (
    df_clean.loc[rimi_mask, "category"]
    .str.rsplit("/", n=1)
    .str[0]
)

display(
    df_clean.loc[rimi_mask, ["category", "category_clean"]]
    .sample(5, random_state=42)
)
```
